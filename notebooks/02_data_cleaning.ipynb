```python
import pandas as pd

runs = pd.read_csv("../data/raw/shift_runs.csv")
machines = pd.read_csv("../data/raw/machines.csv")
shifts = pd.read_csv("../data/raw/shifts.csv", parse_dates=["starts_at"])

machine_lookup = machines[["id", "machine_code", "line_name"]].rename(
    columns={"id": "machine_id"}
)

shift_lookup = shifts[["id", "shift_name", "starts_at"]].rename(
    columns={"id": "shift_id"}
)

data = runs.merge(
    machine_lookup, on="machine_id", how="left", validate="many_to_one"
)
data = data.merge(
    shift_lookup, on="shift_id", how="left", validate="many_to_one"
)

data[["machine_code", "shift_name", "starts_at", "downtime_minutes"]].head()
```

```python
print("Baris sebelum dan sesudah merge : ", len(runs), len(data))

print("Duplikat mesin-shift : ", data.duplicated(["machine_id", "shift_id"]).sum())

print("\nKolom referensi yang kosong : ")
print(data[["machine_code", "shift_name", "starts_at"]].isna().sum())

unit_sesuai = data["good_parts"] + data["scrap_parts"] == data ["total_parts"]
waktu_sesuai = data["run_minutes"] + data["downtime_minutes"] == data["planned_minutes"]

print("\nJumlah baris dengan unit tidak sesuai : ", (~unit_sesuai).sum())
print("Jumlah baris dengan waktu tidak sesuai : ", (~waktu_sesuai).sum())
```
