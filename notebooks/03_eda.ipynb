```python
import pandas as pd

data = pd.read_csv("../data/processed/manufacturing_oee.csv")

total_output = data["total_parts"].sum()
total_ideal = data["ideal_parts"].sum()
total_downtime = data["downtime_minutes"].sum()
total_scrap = data["scrap_parts"].sum()

print(f"Output terhadap ideal : {total_output / total_ideal:.2%}")
print(f"Total downtime : {total_downtime:,} menit")
print(f"Tingkat scrap : {total_scrap / total_output:.2%}")
```

```python
per_mesin = data.groupby("machine_code")[
    ["downtime_minutes", "scrap_parts", "total_parts"]
].sum()

per_mesin["tingkat_scrap_persen"] = (
    100 * per_mesin["scrap_parts"] / per_mesin["total_parts"]
)

per_mesin.sort_values("downtime_minutes", ascending=False).round(2)
```
